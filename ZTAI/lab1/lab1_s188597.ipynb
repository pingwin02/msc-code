```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch

tokenizer = AutoTokenizer.from_pretrained("openai-community/gpt2-large")

model = AutoModelForCausalLM.from_pretrained("openai-community/gpt2-large", torch_dtype=torch.bfloat16, max_length=128)
model = model.to("cuda")

p = pipeline("text-generation", model=model, tokenizer=tokenizer, max_length=128, device="cuda")
```

```python
print(p("""
Q: What color is a sky during sunset?
"""))
```

```python
print(p("""
Q: What color is a sky during day?
A: Blue

Q: What color is a sky during storm?
A: Dark blue

Q: What color is a sky during sunset?
"""))
```

```python
print(p("""
Q: One ball costs 1 dollar. How much cost two balls?
"""))
```

```python
print(p("""
Q: One book costs 1 dollar. How much cost two books?
A: 2 dollars

Q: One bike costs 1 dollar. How much cost two bikes?
A: 2 dollars

Q: One ball costs 1 dollar. How much cost two balls?
"""))
```

```python
print(p("""
Q: There are 5 people in queue. 2 people left and 3 people come. How many people are in queue?
"""))
```

```python
print(p("""
Q: There are 2 apples. One was eaten and two were bought. How many apples are there?
A: 3 apples

Q: There are 3 bananas. Three was eaten and one was bought. How many bananas are there?
A: 1 banana

Q: There are 5 people in queue. 2 people left and 3 people come. How many people are in queue?
"""))
```
