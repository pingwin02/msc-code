````python
import time, requests

API_URL = "http://localhost:8000/v1/completions"
MODEL = "Qwen/Qwen3-4B-Thinking-2507"

payload = {
    "model": MODEL,
    "prompt": "Explain the history of computing in 3 sentences.",
    "max_tokens": 50,
    "temperature": 0.0,
}
headers = {"Content-Type": "application/json"}

requests.post(API_URL, headers=headers, json=payload)

num_requests = 5
total_tokens = 0
res = []

t0_total = time.time()
for i in range(num_requests):
    req_t0 = time.time()
    resp = requests.post(API_URL, headers=headers, json=payload).json()
    req_time = time.time() - req_t0

    tokens = resp["usage"]["completion_tokens"]
    total_tokens += tokens
    res.append(f"Request {i+1} | Latency: {req_time:.3f} s | Tokens generated: {tokens}")

total_time = time.time() - t0_total

res.append("-" * 40)
res.append(f"Total Test Time: {total_time:.3f} s")
res.append(f"Throughput: {total_tokens / total_time:.2f} tokens/s")

print("\n".join(res))
````
